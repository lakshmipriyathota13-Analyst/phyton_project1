```python
#1. Find the length of the list using len().

marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]
for i in range(0,len(marks),1):
 print(marks[i])
```

```python
#2. Find the total marks using sum().

marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]
total_marks=sum(marks)
print(marks)
print(total_marks)
```

```python
#3. Find the highest marks using max().
marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]
total_marks=max(marks)
print(marks)
print(total_marks)
```

```python
#4. Find the lowest marks using min().
marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]
total_marks=min(marks)
print(marks)
print(total_marks)
```

```python
#5. Sort the marks in ascending order.

marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]
marks.sort()
print(marks)
```

```python
#6. Sort the marks in descending order.

marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]
marks.sort(reverse=True)
print(marks)
```

```python
#7. Count how many students scored more than 50.

marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]
count=0
for i in marks:
  if i>50:
   count=count+1
print(count,"students scored more than 50")
```

```python
#8. Count how many students passed, assuming passing marks are 40.

marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]
count=0
for i in marks:
     if i>=40:
      count=count+1
print(count,"students Passed the Exam")
```

```python
#9. Count how many students failed, assuming passing marks are 40.

marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]
count=0
for i in marks:
    if i<40:
        count=count+1
print(count,"students Failed Exam")
```

```python
#10. Calculate the average marks of all students.

marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]

total_marks = sum(marks)

average_marks = total_marks / len(marks)

print("Marks List:", marks)
print("Total Marks =", total_marks)
print("Average Marks =", average_marks)
```
