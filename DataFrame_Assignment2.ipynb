# Filtering – Basic

```python
#1. Display all customers whose Loan_Amount is greater than 500000.
print(df[df["Loan_Amount"]>500000])
```

```python
#2. Display customers whose Interest_Rate is less than 10%.
print(df[df["Interest_Rate"]<10])
```

```python
#3. Display customers whose Credit_Score is greater than 750.
print(df[df["Credit_Score"]>750])
```

```python
#4. Display all customers whose Loan_Status is "Approved".
print(df[df["Loan_Status"]=="Approved"])
```

```python
#5. Display all customers who took a Home loan.
print(df[df["Loan_Type"]=="Home"])
```

```python
#6. Display all customers from Delhi.
print(df[df["City"]=="Delhi"])
```

# Filtering – AND / OR / NOT

```python
#7. Display customers whose Loan_Amount is greater than 500000 AND Credit_Score is greater than 750.
print(df[(df["Loan_Amount"]>500000)& (df["Credit_Score"]>750) ])
```

```python
#8. Display customers from Delhi or Mumbai.
print(df[(df["City"]=="Delhi") |(df["City"]=="Mumbai") ])
```

```python
#9. Display customers who have Credit_Score less than 700 OR Interest_Rate greater than 12%
print(df[(df["Credit_Score"]<700)| (df["Interest_Rate"]>12)])
```

```python
#10. Display customers who are NOT from Delhi.
print(df[df["City"]!="Delhi"])
```

```python
#11. Display customers whose loan is Approved AND Loan_Amount is greater than 600000.

print(df[(df["Loan_Status"]=="Approved")& (df["Loan_Amount"]>600000)])
```

```python
#18.Find the total Loan_Amount for each City.
print(df.groupby("City")["Loan_Amount"].sum())
```

```python
#21.Find the average Interest_Rate for each City.
print(df.groupby("City")["Interest_Rate"].mean())
```

```python
#22.Find the total Loan_Amount for each Loan_Status.
print(df.groupby("Loan_Status")["Loan_Amount"].sum())
```

```python
#23.Find the maximum Credit_Score for each Loan_Type.
print(df.groupby("Loan_Type")["Credit_Score"].max())
```

```python
#24.Find the total Loan_Amount for each City and Loan_Type.
print(df.groupby(["City","Loan_Type"])["Loan_Amount"].sum())
```

```python
# GroupBy with Multiple Aggregations



df.groupby("City")["Loan_Amount"].agg(["sum", "mean", "max", "min"])


```

```python
print(df.groupby("City")["Loan_Amount"].agg(["sum","max","min","mean"]))
```

```python
result=df.groupby("Loan_Type")[["Interest_Rate","Credit_Score","Loan_Amount"]].agg(["mean","sum"])
print(result)
```
