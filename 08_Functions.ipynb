```python
# Create a function to calculate the factorial of a number.
def factorial(n):
   if n==0 or n==1:
     return 1
   else:
      return n*factorial(n-1)

num1=int(input("enter your num"))
print(f"Factorial of {num1} is {factorial(num1)}")
```

# Functions with Lists

```python
#Create a function that accepts a list and returns its length.
def f1(list):
    for i in range(len(list)):
        print(list[i])
L1=[40,85,65,89,71,26,63]
f1(L1)
```

```python
#Create a function that accepts a list and returns the sum of all elements.
def f2(list):
    total=0
    for i in range(len(list)):
      total=total+list[i]
    return total
L2=[5,6,7,8,9,2]   
print(f2(L2))
```

```python
#Create a function that accepts a list and returns the maximum value.
def f3(list):
    return max(list)
L1=[40,85,65,89,71,26,63]

print(f"The highest value in the list is: {(f3(L1))}")
```

```python
#Create a function that accepts a list of marks and returns the number of passed students.
def f4(marks_list):
    pass_count=0
    for marks in marks_list:
      if marks>=40:
        pass_count=pass_count+1
        
      else:
        print("failed")
    return pass_count
stu_marks=[45,69,79,58,95,39,84,33,86,99,23]
count=f4(stu_marks)
print(count)
```

```python
#Create a function that accepts a list and returns a new list containing only even numbers.
def get_even_numbers(list):
    even_list=[]
    for num in list:
        if num%2==0:
            even_list.append(num)
    return even_list

L=[66,79,24,98,86,45,40]
print("Even Numbers are:",get_even_numbers(L))
```

# Functions with Dictionaries

```python
#Create a function that accepts a student dictionary and displays the student's name and marks.
def get_stu_marks(students):
    for names,marks  in students.items():
        print(names,marks)
students={"priya":95,"pradeep":90,"ram":75,"mokshit":81,"asha":39,"shiva":35}
get_stu_marks(students)
```

```python
#Create a function that accepts a dictionary of student marks and returns the highest marks.
def get_max(students):
    top_student=max(students,key=students.get)
    highest_marks=students[top_student]
    print(f"Top Student: {top_student}, Marks: {highest_marks}")
students={"priya":95,"pradeep":90,"ram":75,"mokshit":81,"asha":39,"shiva":35}
get_max(students) 
```

```python
#Create a function that accepts a dictionary of students and marks and returns the number of passed students.
def get_passed_count(students,pass_marks=40):
    passed_count = 0
    for marks in students.values():
        if marks >= pass_marks:
            passed_count += 1
    return passed_count
students={"priya":95,"pradeep":90,"ram":75,"mokshit":81,"asha":39,"shiva":35}
print(get_passed_count(students))
```
