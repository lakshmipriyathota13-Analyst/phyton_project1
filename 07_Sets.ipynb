```python
#Create a set containing 5 numbers.
items=["burger","pizza","doughnut","fish and chips","crumpets","cheescake","fruit boba tea"] 
print(items)
```

```python
# Add a new element using add().
items=["burger","pizza","doughnut","fish and chips","crumpets","cheescake","fruit boba tea"] 
print(items)
items.insert(3,"tarts")
print(items)
```

```python
# Remove an element using remove().
items=["burger","pizza","doughnut","fish and chips","crumpets","cheescake","fruit boba tea"] 
print(items)
items.remove("fruit boba tea")
print(items)
```

```python
#Check whether an element exists in a set.
items=["burger","pizza","doughnut","fish and chips","crumpets","cheescake","fruit boba tea"] 
print(items)
value="pizza"
if value in items:
    print(value)
```

```python
#Create two sets and find their union.
set1={1,2,3,4,5,6,}
set2={5,6,7,8,9,10}
U_set=set1.union(set2)
print(set1)
print(set2)
print(f"The Union Of Two Sets are:{U_set}")
```

```python
#Create two sets and find their intersection.
set1={1,2,3,4,5,6,}
set2={5,6,7,8,9,10}
U_set=set1.intersection(set2)
print(set1)
print(set2)
print(f"The intersection  Of Two Sets are:{U_set}")
```

```python
#Create two sets and find their difference.
set1={1,2,3,4,5,6,}
set2={5,6,7,8,9,10}
U_set=set1.difference(set2)
print(set1)
print(set2)
print(f"The difference between  Two Sets are:{U_set}")
```

```python
#Create a set containing duplicate numbers and observe the result.
numbers={1,2,1,2,3,4,5,6,5,5,7,8,8,9,11,11}
print(f"Set with  duplicate numbers removed:{numbers}")
```

```python
# Given two sets of student names, find students present in both sets.
student_set1={"pradeep","priya","ram","mokshit","nandan","sudha","rani"}
student_set2={"vani","vijaya","rao","pradeep","priya","ram","srikanth"}
common_students=student_set1.intersection(student_set2)
print(student_set1)
print(student_set2)
print(f"The common students present in two sets are:{common_students}")
```
