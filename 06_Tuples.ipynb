```python
# Create a tuple containing 5 numbers and print it.
marks=(85,65,74,89,52,31,60,29,95)
print(marks)
```

```python
# Print the first element of a tuple.
marks=(85,65,74,89,52,31,60,29,95)
print(marks)
print(f"First Element of Tuple is :{marks[0]}")
```

```python
# Print the last element of a tuple.
marks=(85,65,74,89,52,31,60,29,95)
print(marks)
print(f"Last Element of Tuple is :{marks[8]}")
```

```python
#Find the length of a tuple using len().
marks=(85,65,74,89,52,31,60,29,95)
for i in range(1,len(marks),1):
    print(marks[i])
```

```python
#Check whether a particular value exists in a tuple.
marks=(85,65,74,89,52,31,60,29,95)
value=31
if value in marks:
    print(f"Value is present in the tuple :{value}")
```

```python
#Count how many times a value occurs using count().
```

```python
#Find the index of a value using index().
marks=(85,65,74,89,52,31,60,29,95)
print(marks.index(52))
```

```python
#Find the maximum value using max().
marks=(85,65,74,89,52,31,60,29,95)
highest_marks=max(marks)
print(marks)
print(f"Highest Marks is:{highest_marks}")
```

```python
#Find the minimum value using min().
marks=(85,65,74,89,52,31,60,29,95)
lowest_marks=min(marks)
print(marks)
print(f"Lowest Marks is:{lowest_marks}")
```

```python
#Find the total of tuple values using sum().
marks=(85,65,74,89,52,31,60,29,95)
total=sum(marks)
print(marks)
print(total)
```
