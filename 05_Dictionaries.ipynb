```python
#Create a dictionary containing 5 students and their marks. Find the student with the highest marks.
students={"mokshith":85,"bhargav":90,"pradeep":95,"lakshmi":36,"priya":31,"ram":45}
print(students)
topper=max(students,key=students.get)
highest_marks=students[topper]
print(highest_marks)
```

```python
#Create a dictionary containing 5 products and their prices. Find the most expensive product.
products={"laptop":80000,"mobile":30000,"tv":45000,"fridge":60000,"cooler":10000}
top_product=max(products,key=products.get)
expensive=products[top_product]
print("\n---Bill Details---")
print(products)
print(top_product)
print("The Exensive product is=",expensive)
```

```python
#Create a dictionary containing student names and marks. Count how many students passed.
students={"mokshith":85,"bhargav":90,"pradeep":95,"lakshmi":36,"priya":31,"ram":45}
print(students)
pass_mark = 40

# Count how many students passed
passed_count = sum(1 for mark in students.values() if mark >= pass_mark)
print("\n---Student Marks---")
print("Students and Marks:", students)
print("Number of students passed =", passed_count)
```

```python
#Create a dictionary containing student names and marks. Count how many students failed.
students={"mokshith":85,"bhargav":90,"pradeep":95,"lakshmi":36,"priya":31,"ram":45}
print(students)
pass_mark = 40

# Count how many students failed
failed_count = sum(1 for mark in students.values() if mark < pass_mark)
print("\n---Student Marks---")
print("Students and Marks:", students)
print("Number of students Failed =", failed_count)
```

```python
#Create a dictionary containing employee names and salaries. Find the highest salary.
employees={"pradeep":200000,"priya":80000,"ram":40000,"mokshith":350000,"vani":50000,"srikanth":90000,"rao":20000}
sal=max(employees,key=employees.get)
highest_salary=employees[sal]
print("\n---Employees & Salaries---")
print(employees)
print(f"The Highest Salary is:{highest_salary}")
```

```python
#Create a dictionary of products and prices. Calculate the total value of all products.
products={"laptop":80000,"mobile":30000,"tv":45000,"fridge":60000,"cooler":10000}
total_sales=sum(products.values())
print("\n---Sales Details---")
print(products)
print(f"The Total Value of all Products is :{total_sales}")
```
